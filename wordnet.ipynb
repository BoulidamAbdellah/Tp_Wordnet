```python
import nltk

# Téléchargements nécessaires
nltk.download('wordnet')    # pour WordNet
```

```python
# ===================== Imports =====================
from nltk.corpus import wordnet as wn
import networkx as nx
import matplotlib.pyplot as plt

# ===================== Fonction affichage graphe =====================
def afficher_graphe(G, edge_colors, titre):
    plt.figure(figsize=(14,10))
    pos = nx.spring_layout(G, k=0.6, iterations=50, seed=42)
    nx.draw(G, pos, with_labels=True, node_size=2000, node_color='lightblue', arrows=True, edge_color=edge_colors, font_size=10)
    
    # Légende des couleurs
    legend_elements = [
        Line2D([0], [0], color='red', lw=2, label='Hyperonyme'),
        Line2D([0], [0], color='blue', lw=2, label='Hyponyme'),
        Line2D([0], [0], color='green', lw=2, label='Méronyme'),
        Line2D([0], [0], color='orange', lw=2, label='Synonyme'),
        Line2D([0], [0], color='black', lw=2, label='Relation de la phrase')
    ]
    plt.legend(handles=legend_elements, loc='upper right')
    
    plt.title(titre)
    plt.axis('off')
    plt.show()


# ===================== Exercice 1 =====================
mots1 = ["cat", "chased", "mouse", "garden"]
synsets1 = ['cat.n.01', 'chase.v.01', 'mouse.n.01', 'garden.n.01']

print("=== Exercice 1: Synsets et relations ===")
G1 = nx.DiGraph()
colors1 = []

for mot, s in zip(mots1, synsets1):
    syn = wn.synset(s)
    label = s.split('.')[0]
    print(f"\nMot: {mot} -> {s}: {syn.definition()}")
    
    hypers = [h.name() for h in syn.hypernyms()]
    hypos = [h.name() for h in syn.hyponyms()]
    meros = [m.name() for m in syn.part_meronyms()]
    print(f"  Hyperonymes: {hypers}")
    print(f"  Hyponymes: {hypos}")
    print(f"  Méronymes: {meros}")
    
    for h in syn.hypernyms():
        G1.add_edge(label, h.name().split('.')[0]); colors1.append('red')
    for h in syn.hyponyms():
        G1.add_edge(h.name().split('.')[0], label); colors1.append('blue')
    for m in syn.part_meronyms():
        G1.add_edge(label, m.name().split('.')[0]); colors1.append('green')

afficher_graphe(G1, colors1, "Exercice 1 - Graphe sémantique")

# ===================== Exercice 2 =====================
mots2 = ["chef", "prepared", "meal", "served", "guests"]
synsets2 = ['chef.n.01', 'prepare.v.02', 'meal.n.01', 'serve.v.05', 'guest.n.01']

print("\n=== Exercice 2: Synsets et relations ===")
G2 = nx.DiGraph()
colors2 = []

for mot, s in zip(mots2, synsets2):
    syn = wn.synset(s)
    label = s.split('.')[0]
    print(f"\nMot: {mot} -> {s}: {syn.definition()}")
    
    hypers = [h.name() for h in syn.hypernyms()]
    hypos = [h.name() for h in syn.hyponyms()]
    meros = [m.name() for m in syn.part_meronyms()]
    syns = [l.name() for l in syn.lemmas()[:2]]
    
    print(f"  Hyperonymes: {hypers}")
    print(f"  Hyponymes: {hypos}")
    print(f"  Méronymes: {meros}")
    print(f"  Synonymes: {syns}")
    
    for h in syn.hypernyms():
        G2.add_edge(label, h.name().split('.')[0]); colors2.append('red')
    for h in syn.hyponyms()[:2]:
        G2.add_edge(h.name().split('.')[0], label); colors2.append('blue')
    for m in syn.part_meronyms():
        G2.add_edge(label, m.name().split('.')[0]); colors2.append('green')
    for l in syn.lemmas()[:2]:
        if l.name() != label:
            G2.add_edge(l.name(), label); colors2.append('orange')

# Relations de la phrase
G2.add_edge('chef', 'prepare'); colors2.append('black')
G2.add_edge('prepare', 'meal'); colors2.append('black')
G2.add_edge('serve', 'meal'); colors2.append('black')
G2.add_edge('serve', 'guest'); colors2.append('black')

afficher_graphe(G2, colors2, "Exercice 2 - Graphe sémantique")

# ===================== Exercice 3 =====================
mots3 = ["Alice", "buy", "laptop", "store", "write", "report"]
synsets3 = ['buy.v.01', 'laptop.n.01', 'shop.n.01', 'write.v.01', 'report.n.01']

print("\n=== Exercice 3: Synsets et relations ===")
G3 = nx.DiGraph()
colors3 = []

for mot, s in zip(mots3[1:], synsets3):
    syn = wn.synset(s)
    label = s.split('.')[0]
    print(f"\nMot: {mot} -> {s}: {syn.definition()}")
    
    hypers = [h.name() for h in syn.hypernyms()]
    hypos = [h.name() for h in syn.hyponyms()]
    meros = [m.name() for m in syn.part_meronyms()]
    
    print(f"  Hyperonymes: {hypers}")
    print(f"  Hyponymes: {hypos}")
    print(f"  Méronymes: {meros}")
    
    for h in syn.hypernyms():
        G3.add_edge(label, h.name().split('.')[0]); colors3.append('red')
    for h in syn.hyponyms()[:2]:
        G3.add_edge(h.name().split('.')[0], label); colors3.append('blue')
    for m in syn.part_meronyms():
        G3.add_edge(label, m.name().split('.')[0]); colors3.append('green')

# Relations de la phrase
G3.add_edge('Alice', 'buy'); colors3.append('black')
G3.add_edge('buy', 'laptop'); colors3.append('black')
G3.add_edge('buy', 'shop'); colors3.append('black')
G3.add_edge('laptop', 'write'); colors3.append('black')
G3.add_edge('write', 'report'); colors3.append('black')

afficher_graphe(G3, colors3, "Exercice 3 - Graphe sémantique")

# ===================== Exercice 4 =====================
mots4 = ["car", "bus", "bicycle", "vehicle", "machine", "engine", "battery", "fuel", "passenger", "road", "airplane", "sky"]
synsets4 = ['car.n.01', 'bus.n.01', 'bicycle.n.01', 'vehicle.n.01', 'machine.n.01', 'engine.n.01', 
            'battery.n.02', 'fuel.n.01', 'passenger.n.01', 'road.n.01', 'airplane.n.01', 'sky.n.01']

print("\n=== Exercice 4: Synsets et relations ===")
G4 = nx.DiGraph()
colors4 = []

for mot, s in zip(mots4, synsets4):
    syn = wn.synset(s)
    label = s.split('.')[0]
    print(f"\nMot: {mot} -> {s}: {syn.definition()}")
    
    hypers = [h.name() for h in syn.hypernyms()]
    hypos = [h.name() for h in syn.hyponyms()]
    meros = [m.name() for m in syn.part_meronyms()]
    
    print(f"  Hyperonymes: {hypers}")
    print(f"  Hyponymes: {hypos}")
    print(f"  Méronymes: {meros}")
    
    for h in syn.hypernyms():
        G4.add_edge(label, h.name().split('.')[0]); colors4.append('red')
    for h in syn.hyponyms()[:1]:
        G4.add_edge(h.name().split('.')[0], label); colors4.append('blue')
    for m in syn.part_meronyms():
        G4.add_edge(label, m.name().split('.')[0]); colors4.append('green')

# Relations supplémentaires du texte
G4.add_edge('car', 'engine'); colors4.append('green')
G4.add_edge('bus', 'engine'); colors4.append('green')
G4.add_edge('car', 'battery'); colors4.append('green')
G4.add_edge('vehicle', 'road'); colors4.append('black')
G4.add_edge('airplane', 'sky'); colors4.append('black')

afficher_graphe(G4, colors4, "Exercice 4 - Graphe sémantique")
```
